```python
import requests
from bs4 import BeautifulSoup

# Fixed webpage URL
url = "https://www.python.org/"

# Send request to webpage
response = requests.get(url)

# Check whether webpage was successfully accessed
if response.status_code == 200:

    # Parse webpage HTML
    soup = BeautifulSoup(response.text, "html.parser")

    # Extract title
    title = soup.title.string.strip()

    # Save title to a text file
    with open("webpage_title.txt", "w") as file:
        file.write("Webpage URL: " + url + "\n")
        file.write("Webpage Title: " + title + "\n")

    print("Automation completed successfully!")
    print("Webpage Title:", title)
    print("Saved to: webpage_title.txt")

else:
    print("Unable to access the webpage.")
    print("Status Code:", response.status_code)
```

## How the Automation Works

1. The program stores the webpage URL.
2. The requests library sends a request to the webpage.
3. The program checks the response status.
4. BeautifulSoup reads the HTML content.
5. The program finds the webpage title using the `<title>` tag.
6. The extracted title is automatically saved into a text file.
7. The user does not need to manually open the webpage or copy the title.
